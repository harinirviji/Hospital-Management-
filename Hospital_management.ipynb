```python


import numpy as np

patients = []

# Display Menu
def display_menu():
    print("\n===================================")
    print("       HOSPITAL MANAGEMENT SYSTEM")
    print("===================================")
    print("1. Add Patient")
    print("2. View All Patients")
    print("3. Search Patient")
    print("4. Update Patient")
    print("5. Delete Patient")
    print("6. Calculate Hospital Bill")
    print("7. Hospital Data Analysis")
    print("8. Department Analysis")
    print("9. Exit")
    print("===================================")


# 1. Add Patient
def add_patient():
    print("\n--- ADD PATIENT ---")

    patient_id = input("Enter Patient ID: ")

    for patient in patients:
        if patient["id"] == patient_id:
            print("Patient ID already exists!")
            return

    name = input("Enter Patient Name: ")

    age = int(input("Enter Patient Age: "))

    gender = input("Enter Gender: ")

    department = input("Enter Department: ")

    days = int(input("Enter Number of Days: "))

    bill = float(input("Enter Bill Amount: "))

    if age <= 0 or days < 0 or bill < 0:
        print("Invalid input! Please try again.")
        return

    patient = {
        "id": patient_id,
        "name": name,
        "age": age,
        "gender": gender,
        "department": department,
        "days": days,
        "bill": bill
    }

    patients.append(patient)

    print("\nPatient Added Successfully!")


# 2. View All Patients
def view_patients():
    print("\n--- ALL PATIENT RECORDS ---")

    if len(patients) == 0:
        print("No Patient Records Found!")
        return

    for patient in patients:
        print("\nPatient ID:", patient["id"])
        print("Name:", patient["name"])
        print("Age:", patient["age"])
        print("Gender:", patient["gender"])
        print("Department:", patient["department"])
        print("Hospital Days:", patient["days"])
        print("Bill Amount: Rs.", patient["bill"])


# 3. Search Patient
def search_patient():
    print("\n--- SEARCH PATIENT ---")

    patient_id = input("Enter Patient ID: ")

    for patient in patients:
        if patient["id"] == patient_id:
            print("\nPatient Found!")
            print("Patient ID:", patient["id"])
            print("Name:", patient["name"])
            print("Age:", patient["age"])
            print("Gender:", patient["gender"])
            print("Department:", patient["department"])
            print("Bill Amount:", patient["bill"])
            return

    print("Patient Not Found!")


# 4. Update Patient
def update_patient():
    print("\n--- UPDATE PATIENT ---")

    patient_id = input("Enter Patient ID: ")

    for patient in patients:
        if patient["id"] == patient_id:

            print("\n1. Update Name")
            print("2. Update Age")
            print("3. Update Gender")
            print("4. Update Department")
            print("5. Update Hospital Days")
            print("6. Update Bill")

            choice = input("Enter Choice: ")

            if choice == "1":
                patient["name"] = input("Enter New Name: ")

            elif choice == "2":
                age = int(input("Enter New Age: "))
                if age <= 0:
                    print("Invalid age!")
                    return
                patient["age"] = age

            elif choice == "3":
                patient["gender"] = input("Enter New Gender: ")

            elif choice == "4":
                patient["department"] = input("Enter New Department: ")

            elif choice == "5":
                days = int(input("Enter New Days: "))
                if days < 0:
                    print("Invalid days!")
                    return
                patient["days"] = days

            elif choice == "6":
                bill = float(input("Enter New Bill: "))
                if bill < 0:
                    print("Invalid bill!")
                    return
                patient["bill"] = bill

            else:
                print("Invalid Choice!")
                return

            print("Patient Updated Successfully!")
            return

    print("Patient Not Found!")


# 5. Delete Patient
def delete_patient():
    print("\n--- DELETE PATIENT ---")

    patient_id = input("Enter Patient ID: ")

    for patient in patients:
        if patient["id"] == patient_id:
            patients.remove(patient)
            print("Patient Deleted Successfully!")
            return

    print("Patient Not Found!")


# 6. Calculate Hospital Bill
def calculate_bill():
    print("\n--- HOSPITAL BILL CALCULATION ---")

    patient_id = input("Enter Patient ID: ")

    for patient in patients:
        if patient["id"] == patient_id:

            consultation_fee = 500
            room_charge = patient["days"] * 1000
            total_bill = consultation_fee + room_charge

            patient["bill"] = total_bill

            print("\nPatient Name:", patient["name"])
            print("Consultation Fee: Rs.", consultation_fee)
            print("Room Charge: Rs.", room_charge)
            print("Total Hospital Bill: Rs.", total_bill)

            return

    print("Patient Not Found!")


# 7. Hospital Data Analysis Using NumPy
def hospital_analysis():
    print("\n--- HOSPITAL DATA ANALYSIS ---")

    if len(patients) == 0:
        print("No Data Available!")
        return

    ages = np.array([p["age"] for p in patients])
    bills = np.array([p["bill"] for p in patients])
    days = np.array([p["days"] for p in patients])

    print("Total Patients:", len(patients))
    print("Average Patient Age:", np.mean(ages))
    print("Youngest Patient Age:", np.min(ages))
    print("Oldest Patient Age:", np.max(ages))

    print("Total Hospital Revenue: Rs.", np.sum(bills))
    print("Average Hospital Bill: Rs.", np.mean(bills))
    print("Highest Hospital Bill: Rs.", np.max(bills))
    print("Lowest Hospital Bill: Rs.", np.min(bills))

    print("Total Hospital Days:", np.sum(days))


# 8. Department Analysis
def department_analysis():
    print("\n--- DEPARTMENT ANALYSIS ---")

    if len(patients) == 0:
        print("No Data Available!")
        return

    departments = np.array(
        [p["department"] for p in patients]
    )

    unique_departments, counts = np.unique(
        departments, return_counts=True
    )

    for department, count in zip(
        unique_departments, counts
    ):
        print(department, ":", count, "Patients")


# Main Program
while True:

    display_menu()

    choice = input("Enter Your Choice: ")

    if choice == "1":
        add_patient()

    elif choice == "2":
        view_patients()

    elif choice == "3":
        search_patient()

    elif choice == "4":
        update_patient()

    elif choice == "5":
        delete_patient()

    elif choice == "6":
        calculate_bill()

    elif choice == "7":
        hospital_analysis()

    elif choice == "8":
        department_analysis()

    elif choice == "9":
        print("\nThank You for Using Hospital Management System!")
        print("Program Exited Successfully.")
        break

    else:
        print("Invalid Choice! Please Try Again.")
```
